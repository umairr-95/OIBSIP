```python
# 8.Outlier Detection: use IQR method or Z-score to identify outliers in numeric columns; decide and document whether to cap, remove, or retain each
duration_parts = df["duration"].str.extract(r"^\s*(\d+)\s*(min|mins|Season|Seasons)\s*$")
df["duration_value"] = pd.to_numeric(duration_parts[0],errors="coerce")
df["duration_unit"] = duration_parts[1].replace({"mins": "min","Seasons": "Season"})
display(df[["duration", "duration_value", "duration_unit"]].head(10))
```

```python
# IQR FOR DURATION VALUE
Q1 = df["duration_value"].quantile(0.25)
Q3 = df["duration_value"].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["duration_value"] < lower_bound) |
    (df["duration_value"] > upper_bound)
]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
#decesion of whether to cap, remove, or retain each.
#The IQR method was used to identify potential outliers. The detected values were reviewed as potential domain-valid observations. 
#Extreme release years and long movie durations can represent legitimate Netflix titles rather than data-entry errors. 
#Therefore, the outliers were retained instead of being capped or removed. This prevents loss of valid information and avoids distorting the dataset.
```

```python
# Data type correction: ensure all columns have the correct dtype (e.g., dates as datetime, IDs as string, monetary values as float)

summary = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Duplicate Count",
        "Null Count"
    ],
    "Before Cleaning": [
        len(df),
        int(df.duplicated().sum()),
        int(df.isnull().sum().sum())
    ],
    "After Cleaning": [
        len(df),
        int(df.duplicated().sum()),
        int(df.isnull().sum().sum())
    ]
})

display(summary)
```
